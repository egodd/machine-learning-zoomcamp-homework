## question 7

```python
short_df = df[df['origin'] == 'Asia'][['vehicle_weight', 'model_year']].head(7)
short_df
```

```python
# underlying numpy array
X = np.array(short_df)
X, X.T
```

```python
X.shape, X.T.shape
```

```python
# transpose of X times X (dot product)
XTX = X.T.dot(X)
XTX
```

```python
y = [1100, 1300, 800, 900, 1000, 1100, 1200]
```

```python
# get the inverse of XTX
z =np.linalg.inv(XTX)
z
```

```python
#  multiply the inverse of XTX with X.T then multiply by y
product = z.dot(X.T).dot(y)
product
```
